```python
import os
from pathlib import Path
from utils import data_loader_utils
from utils import user_defined_functions as udf
import itertools 
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.mlab as mlab
import pandas as pd
import math
from scipy import ndimage
from scipy.stats import pearsonr
import scipy.signal as signal
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.stattools import adfuller
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.statespace.sarimax import SARIMAX
```

```python
file_good=Path('./data/M01/OP07/good/M01_Aug_2019_OP07_000.h5')
file_bad=Path('./data/M01/OP07/bad/M01_Aug_2019_OP07_000.h5')

data_good=data_loader_utils.datafile_read(file_good)
data_bad=data_loader_utils.datafile_read(file_bad)
```

```python
# --- Plotting Power Spectral Density (PSD) ---

# Parameters
fs = 2000       # Sampling rate (Hz) [cite: 66, 90]
nfft = 1024     # Number of data points used in each block for the FFT

plt.figure(figsize=(12, 6))

# 1. Plot PSD for Good Data (X-axis is column 0)
# We use column index 0 assuming good_data is a numpy array or dataframe
# If it's a dataframe, use .iloc[:, 0] or the column name
plt.psd(data_good[:, 0], NFFT=nfft, Fs=fs, color='blue', alpha=0.7, label='OK (Healthy)')

# 2. Plot PSD for Bad Data
plt.psd(data_bad[:, 0], NFFT=nfft, Fs=fs, color='red', alpha=0.7, label='NOK (Faulty)')

# Add markers for Spindle Speed harmonics (200 Hz, 400 Hz, etc.)
for harmonic in [200, 400, 600, 800]:
    plt.axvline(x=harmonic, color='green', linestyle='--', alpha=0.5)

plt.title('Power Spectral Density: OK vs NOK (X-Axis)')
plt.legend()
plt.xlim(0, 1000) # Nyquist limit is 1000 Hz
plt.show()
```

```python
# --- CELL: Stationarity & Autocorrelation (Corrected for NumPy) ---

# Configuration
target_axis = 0  # 0 = X-axis
rotation_period = 10 # 2000Hz / 200Hz = 10 samples

# 1. Stationarity Check (ADF Test)
def check_stationarity(timeseries, title):
    # Handle NaNs (NumPy doesn't have .dropna(), so we check for NaNs manually)
    if np.isnan(timeseries).any():
        timeseries = timeseries[~np.isnan(timeseries)]
    
    # Test on first 5000 samples for speed
    result = adfuller(timeseries[:10000], autolag='AIC')
    
    print(f"\n--- ADF Test: {title} ---")
    print(f"  ADF Statistic: {result[0]:.4f}")
    print(f"  p-value: {result[1]:.4f}")
    
    if result[1] < 0.05:
        print("  -> Result: Stationary (Stable)")
    else:
        print("  -> Result: Non-Stationary (Unstable/Drifting)")

# Select the data using NumPy slicing [:, column_index]
# We assume data_good is shape (N_samples, 3_axes)
x_good = data_good[:, target_axis]
x_bad = data_bad[:, target_axis]

# Run the test
print("Checking Stationarity...")
check_stationarity(x_good, "OK Data (X-axis)")
check_stationarity(x_bad, "NOK Data (X-axis)")

# 2. Autocorrelation Plot (ACF)
fig, axes = plt.subplots(2, 1, figsize=(12, 8), sharex=True)

# Plot OK ACF
plot_acf(x_good, lags=100, ax=axes[0], title='Autocorrelation: OK Data')
# Add green lines for expected rotation (every 10 samples)
for i in range(rotation_period, 101, rotation_period):
    axes[0].axvline(x=i, color='green', linestyle='--', alpha=0.5)

# Plot NOK ACF
plot_acf(x_bad, lags=100, ax=axes[1], title='Autocorrelation: NOK Data')

plt.xlabel('Lags (Samples)')
plt.show()
```

```python
# --- CELL: AR Model & Residual Analysis ---

# 1. Setup Data (Extract X-axis from the numpy arrays)
# We assume data_good is shape (N, 3). We take column 0.
x_good = data_good[:, 0]
x_bad = data_bad[:, 0]

# 2. Define Model Parameters
# p=10 corresponds to the 10-sample rotation period (Physics-based)
p_order = 40 

print(f"Training AR({p_order}) model on OK data...")

# 3. Train the Model on Healthy Data
# We use a subset (first 5000 points) to make it fast
train_subset = x_good[:5000]

# ARIMA(p,0,0) is equivalent to an AR(p) model
model = ARIMA(train_subset, order=(p_order, 0, 0))
model_fit = model.fit()

# Print summary to see if the lags are significant
# print(model_fit.summary()) 

# 4. Get Residuals (Prediction Errors)
# A. For the Healthy Data (Self-check)
residuals_ok = model_fit.resid

# B. For the Faulty Data (Anomaly check)
# We apply the *trained* parameters to the faulty data
model_nok = model_fit.apply(x_bad[:5000])
residuals_nok = model_nok.resid

# 5. Visual Comparison
fig, ax = plt.subplots(2, 1, figsize=(12, 8), sharey=True)

# Plot OK Residuals
ax[0].plot(residuals_ok, color='blue', alpha=0.6)
ax[0].set_title(f"Residuals - OK Data (MSE: {np.mean(residuals_ok**2):.2f})")
ax[0].set_ylabel("Error Amplitude")
ax[0].grid(True)

# Plot NOK Residuals
ax[1].plot(residuals_nok, color='red', alpha=0.8)
ax[1].set_title(f"Residuals - NOK Data (MSE: {np.mean(residuals_nok**2):.2f})")
ax[1].set_ylabel("Error Amplitude")
ax[1].set_xlabel("Time (Samples)")
ax[1].grid(True)

plt.tight_layout()
plt.show()

# 6. Final Quantitative Score
mse_ok = np.mean(residuals_ok**2)
mse_nok = np.mean(residuals_nok**2)
ratio = mse_nok / mse_ok

print("\n--- Anomaly Detection Results ---")
print(f"Mean Squared Error (OK):  {mse_ok:.4f}")
print(f"Mean Squared Error (NOK): {mse_nok:.4f}")
print(f"Anomaly Ratio: {ratio:.2f}")
print("Interpretation: The model's error is {:.1f} times higher on the faulty data.".format(ratio))
```

```python
# --- CELL: Diagnostic Check (Is it White Noise?) ---

# 1. Visual Inspection
plt.figure(figsize=(12, 4))
plt.plot(residuals_ok, color='blue', alpha=0.5)
plt.title("Residuals of OK Data (Should look random)")
plt.ylabel("Error")
plt.xlabel("Time")
plt.show()

# 2. Statistical Inspection (ACF of Residuals)
fig, ax = plt.subplots(figsize=(10, 5))
plot_acf(residuals_ok, lags=40, ax=ax, title='ACF of OK Residuals (Ideally all zero)')
plt.show()

# 3. Distribution Inspection (Histogram)
plt.figure(figsize=(10, 5))
plt.hist(residuals_ok, bins=50, color='blue', alpha=0.7, density=True)
plt.title("Histogram of Residuals (Should look Gaussian)")
plt.show()

# 2. Statistical Inspection (ACF of Residuals)
fig, ax = plt.subplots(figsize=(10, 5))
plot_acf(residuals_nok, lags=40, ax=ax, title='ACF of NOK Residuals')
plt.show()

# 3. Distribution Inspection (Histogram)
plt.figure(figsize=(10, 5))
plt.hist(residuals_nok, bins=50, color='blue', alpha=0.7, density=True)
plt.title("Histogram of Residuals")
plt.show()
```
